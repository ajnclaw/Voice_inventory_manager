```python
import sqlite3
import pandas as pd

# Reading while agent_server.py is running is fine -- SQLite allows
# concurrent readers, no need to stop the server first.
conn = sqlite3.connect("inventory.db")
```

```python
# Current catalog -- the cached view (see agent/inventory_db.py)
pd.read_sql_query("SELECT * FROM items ORDER BY category, name", conn)
```

```python
# The real source of truth -- every sale/restock/adjustment, in order.
# current_quantity on items is always just the sum of these per item_id.
pd.read_sql_query("SELECT * FROM transactions ORDER BY created_at DESC", conn)
```

```python
# Transaction history for one specific item, joined with its name
pd.read_sql_query(
    """
    SELECT t.*, i.name
    FROM transactions t
    JOIN items i ON i.id = t.item_id
    WHERE i.name LIKE '%flywheel%'
    ORDER BY t.created_at
    """,
    conn,
)
```
